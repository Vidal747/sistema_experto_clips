Este notebook utiliza **CLIPSPY** para ejecutar cuatro reglas y recomendar una estrategia según la proximidad del examen, la dificultad y el tiempo disponible.

## 2. Motor de inferencia
La función carga `reglas_estudio.clp`, inserta el hecho seleccionado, ejecuta la agenda de CLIPS y devuelve todas las recomendaciones activadas.

```python
from sistema_experto import recomendar

recomendar("muy-cercano", "alta", "poco")
```

## 3. Interfaz

```python
import ipywidgets as widgets
from IPython.display import display

proximidad = widgets.Dropdown(
    options=[("Muy cercano", "muy-cercano"), ("Cercano", "cercano"), ("Lejano", "lejano")],
    description="Examen:",
    style={"description_width": "120px"},
)
dificultad = widgets.Dropdown(
    options=[("Alta", "alta"), ("Baja", "baja")],
    description="Dificultad:",
    style={"description_width": "120px"},
)
tiempo = widgets.Dropdown(
    options=[("Poco", "poco"), ("Mucho", "mucho")],
    description="Tiempo disponible:",
    style={"description_width": "120px"},
)
boton = widgets.Button(description="Obtener recomendación", button_style="primary")
salida = widgets.Output()

def consultar(_):
    with salida:
        salida.clear_output()
        resultados = recomendar(proximidad.value, dificultad.value, tiempo.value)
        if resultados:
            print("Recomendación del sistema experto:")
            for item in resultados:
                print(f"• {item['regla']}: {item['mensaje']}")
        else:
            print("No existe una recomendación para esta combinación de datos.")

boton.on_click(consultar)
display(widgets.VBox([
    widgets.HTML("<h3>Asistente para planificar el estudio</h3><p>Seleccioná las condiciones actuales.</p>"),
    proximidad, dificultad, tiempo, boton, salida
]))
```

## 4. Prueba de una combinación con dos reglas
Con examen cercano, dificultad baja y poco tiempo se activan R02 y R04 porque ambas condiciones son verdaderas.

```python
recomendar("cercano", "baja", "poco")
```
